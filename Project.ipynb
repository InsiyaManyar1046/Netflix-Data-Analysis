```python
import pandas as pd
import matplotlib.pyplot as plt 
df=pd.read_csv('netflix_titles.csv.zip')
#to clean data
df=df.dropna(subset=["type","country","release_year","rating","duration"])
#bar plot of number of movies vs TV shows on netflix 
type_counts=df["type"].value_counts()
plt.figure(figsize=(5,4))
plt.bar(type_counts.index, type_counts.values, color=["skyblue","orange"])
plt.title("Number of movies vs TV shows on netflix")
plt.xlabel("Type")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
#pie chart that shows the percentage of content rating 
rating_count=df["rating"].value_counts()
plt.figure(figsize=(6,5))
plt.pie(rating_count, labels=rating_count.index, autopct="%1.1f%%", startangle=90)
plt.title("Percentage of content rating ")
plt.tight_layout()
plt.show()
```

```python
#histogram to represent distribution od movie duration 
movie_df = df[df["type"]== "Movie"].copy()
movie_df["duration_int"] = movie_df["duration"].str.replace("min","").astype("int")

plt.figure(figsize=(7,6))
plt.hist(movie_df["duration_int"], bins=30, edgecolor="black", color="pink")
plt.title("Distribution od movie duration ")
plt.xlabel("Duration in minutes")
plt.ylabel("Number of movies")
plt.tight_layout()
plt.show()
```

```python
#Scatter plot of released year vs number of shows
release_count=df["release_year"].value_counts().sort_index()
plt.figure(figsize=(7,6))
plt.scatter(release_count.index, release_count.values, marker="^", label="shows in per year")
plt.title("Number of shows in each released year")
plt.xlabel("year")
plt.ylabel("Number of shows")
plt.grid(linestyle="-", color="pink")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Between 1-10 country name check who is having more shows
country_count=df["country"].value_counts().head(10)
plt.figure(figsize=(10,6))
plt.bar(country_count.index, country_count.values,color="skyblue")
plt.title("Top 10 countries by number of shows")
plt.xlabel("Country")
plt.ylabel("Number of shows")
plt.tight_layout()
plt.show()
```

```python
#Subplot and grouping 
#Comparision of movies and tv shows released per year
content_by_year= df.groupby(["release_year","type"]).size().unstack().fillna(0)
fig ,ax=plt.subplots(1,2,figsize=(10,5))
ax[0].plot(content_by_year.index, content_by_year["Movie"], color="red")
ax[0].set_title("Movie released per year")
ax[0].set_xlabel("Years")
ax[0].set_ylabel("Movies")

ax[1].plot(content_by_year.index, content_by_year["TV Show"], color="blue")
ax[0].set_title("Tv shows released per year")
ax[0].set_xlabel("Years")
ax[0].set_ylabel("Tv shows")
```
